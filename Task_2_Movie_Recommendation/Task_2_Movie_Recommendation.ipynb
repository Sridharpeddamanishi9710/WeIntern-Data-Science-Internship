```python
## We want only the "name" value from each dictionary.

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
#Testing 

convert(movies.loc[0, 'genres'])
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
# Extracting the Cast

def convert_cast(obj):
    L = []
    for i in ast.literal_eval(obj):
        if len(L) < 3:
            L.append(i['name'])
        else:
            break
    return L
```

```python
#Testing

convert_cast(movies.loc[0, 'cast'])
```

```python
#Applying to all movies

movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies[['title', 'cast']].head()
```

```python
#Extracting the director

def fetch_director(obj):
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            return i['name']
    return ''
```

```python
#Test
fetch_director(movies.loc[0, 'crew'])
```

```python
#Applying director function to all Movies

movies['director'] = movies['crew'].apply(fetch_director)
```

```python
#selecting only the columns we need
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'director']]
```

```python
movies.head()
```

```python
#convert the lists to strings
movies["genres"] = movies["genres"].apply(lambda x: " ".join(x))
movies["keywords"] = movies["keywords"].apply(lambda x: " ".join(x))
movies["cast"] = movies["cast"].apply(lambda x: " ".join(x))
```

```python
movies[["title", "genres", "keywords", "cast", "director"]].head()
```

```python
#Combine the features into one column called tags
movies['tags'] = (
    movies['overview'] + ' ' +
    movies['genres'] + ' ' +
    movies['keywords'] + ' ' +
    movies['cast'] + ' ' +
    movies['director']
)
```

```python
movies[['title', 'tags']].head()
```

```python
movies['tags'].str.len().describe()
```

```python
#Coverting tags to lowercase before apply to TF-IDF 
movies['tags'] = movies['tags'].str.lower()
```

```python
movies[['title', 'tags']].head()
```

```python
#Checking missing tags
movies['tags'].isnull().sum()
```

```python
# TF-IDF = Term Frequency–Inverse Document Frequency.
# It converts our text into numerical vectors so that we can mathematically compare movies.
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
#Removing common english stop words
tfidf = TfidfVectorizer(stop_words='english')
```

```python
# Transforming the tags
tfidf_matrix = tfidf.fit_transform(movies['tags'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#calculating cosine_similarity becuase it measures how similar two movie vectors are.

similarity = cosine_similarity(tfidf_matrix)
```

```python
similarity.shape
```

```python
#First movie's similarity's scores
similarity[0]
```

```python
movie_indices = pd.Series(movies.index, index=movies['title']).drop_duplicates()
```

```python
# Creating Recommendation function

def recommend(movie):
    if movie not in movie_indices:
        print("Movie not found. Please enter a valid movie title.")
        return
    
    movie_index = movie_indices[movie]
    
    similarity_scores = list(enumerate(similarity[movie_index]))
    
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )
    
    top_movies = similarity_scores[1:6]
    
    print(f"\nTop 5 recommendations for '{movie}':\n")
    
    for rank, (index, score) in enumerate(top_movies, start=1):
        print(f"{rank}. {movies.iloc[index]['title']} → {score:.3f}")
```

```python
recommend("Avatar")
```

```python
recommend("The Dark Knight Rises")
```

```python
recommend("Titanic")
```

```python
recommend("The Avengers")
```

```python
recommend("xyz")
```

```python
movie_name = input("Enter a movie name: ")
recommend(movie_name)
```

```python
## Conclusion

A content-based movie recommendation system was developed using the TMDB 5000 Movie Dataset. Movie information such as overview, genres, keywords, cast, and director was combined to create a unified content feature. TF-IDF was used to convert the textual information into numerical vectors, and cosine similarity was used to identify movies with similar content. The system successfully provides the top 5 movie recommendations based on a user's selected movie and handles invalid movie titles appropriately.
```
